```python
from PIL import Image
import os

# Folder containing PNG images
image_folder = "./"  # Change to your directory if needed

# Get all PNG files in the folder, sorted by name
png_files = sorted([f for f in os.listdir(image_folder) if f.lower().endswith(".png")])

# Open all images and convert to RGB
images = [Image.open(os.path.join(image_folder, f)).convert("RGB") for f in png_files]

# Save as a single PDF
if images:
    output_path = "merged_output.pdf"
    images[0].save(output_path, save_all=True, append_images=images[1:])
    print(f"PDF saved as '{output_path}'")
else:
    print("No PNG files found.")
```
